```python
from langchain_mistralai import ChatMistralAI
```

```python
import os
```

```python
from getpass import getpass
```

```python
os.environ["MISTRAL_API_KEY"]= getpass("Enter your api key:")
```

```python
llm = ChatMistralAI(model = "mistral-large-latest")
```

```python
utr_numbers = ["SBI101","SBI102","PNB103","HPGB104","SBI105","PNB106","HPGB107","HPGB108","PNB109","ICICI110"]
senders_names = ["Rahul","Ritika","Gaurav","Payal","Raghav","Aksham","Komal","Kartik","Akshay","Rupali"]
receivers_names = ["Rohit","Vanshika","Aditi","Kajal","Mohit","Vijay","Pranav","Poorav","Ekta","Isha"]
money = [20000,15000,25000,10000,20000,5000,15000,18000,15000,20000]
```

```python
utr_history = []
```

```python
while True:
   question = input("Enter utr number:")
   if question.strip().lower() == "exit":
     print("Happy to help you in future")
     break
   elif question not in utr_numbers:
    print("UTR not matched! please check UTR number.")
   elif question in utr_numbers:
     utr_history.append(question)
     index = utr_numbers.index(question)
     if len(utr_history) > 3:
      del utr_history[:-3]
     print(utr_history)
     first_question = input("Enter did money deduct from your account(yes/no):")
     second_question = input("Enter did you receive money in your account(yes/no):")
     if first_question.lower().strip() == "no" and second_question.lower().strip() == "no":
        print("There is an error occured in transaction! Try again after some time.")
     elif first_question.lower().strip() == "yes" and second_question.lower().strip() == "no":
        print("Your money will refund in your account within two days.")
     elif first_question.lower().strip() == "yes" and second_question.lower().strip() == "yes":
        print("TRANSACTION COMPLETE!")

     prompt = f"""You are a Paytm machine. Users enters the UTR id's from above given data of {utr_numbers},{senders_names},{receivers_names},{money}
     convert these list into dataset.Then you have to check that either that utr number exists in given data or not.Match properlythe utr number: {question}.
     Don't show all these stepts in the output.Only show important infomation like sender and receiver name.Ask a question to sender that Enter:
     First Question: {first_question}
     Second Question: {second_question}."""
     response = llm.invoke(prompt)
     print(response.content)
     print("*************************************************************")
```
