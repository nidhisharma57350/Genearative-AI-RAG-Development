```python
from langchain_mistralai import ChatMistralAI
```

```python
import os
```

```python
from getpass import getpass
```

```python
os.environ["MISTRAL_API_KEY"] = getpass("Enter your api key:")
```

```python
llm = ChatMistralAI(model = "mistral-large-latest")
```

```python
user_chat_history_list = []
```

```python
while True:
  question = input("Enter your question:")
  if question.strip().lower() == "exit":
    print("Happy to help you in future.")
    break
  prompt = f"""you are a manager of the company which deels with the AC services. And if customes faces any problem then it is your responsibility to
  deal with that problems. firstly one of the most important question you have to asked from the customer that is when you buy this product from our
  company. Do you have it's warrenty card. If the time is more then 3 years then sorry we can't help you.And if the time is less then 3 years then it's
  our company's responsibility to repare your AC.So, tell me what is the problem.This is for user Write your problem in the question {question}. one
  another thing that you have to kept in mind that user's chat history as you are manager of the company then it is your responsiblity to keep previous
  questions also in memory :{user_chat_history_list}"""
  response = llm.invoke(prompt)
  print(response.content)
  user_chat_history = question + response.content
  print(user_chat_history)
  user_chat_history_list.append(user_chat_history)
  if len(user_chat_history_list) > 3:
    del user_chat_history_list[ :-3]
  print(user_chat_history_list)
  print("*****************************************")
```
