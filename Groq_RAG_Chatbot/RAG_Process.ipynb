```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
uploading=files.upload() # files - module,upload - function
```

```python
pdf_file=next(iter(uploading))
print(pdf_file)
```

```python
loader = PyPDFLoader(pdf_file)
print(loader)
```

```python
document=loader.load()
print(document)
```

```python
split=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
```

```python
chunks=split.split_documents(document)
```

```python
print(type(chunks))
```

```python
print(len(chunks))
print(chunks[0])
print(type(chunks[0]))
```

```python
print(chunks)
```

```python
os.environ["GROQ_API_KEY"] = getpass("Enter your groq api key")
```

```python
from langchain_groq import ChatGroq
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
embeddings = HuggingFaceEmbeddings(model_name = "sentence-transformers/all-MiniLM-L6-v2")
```

```python
from langchain_chroma import Chroma
```

```python
vectordb = Chroma.from_documents(documents=chunks,embedding=embeddings)
```

```python
retrieve = vectordb.as_retriever(search_kwargs={"k":5}) # as_retriever- a method.
```

```python
llm = ChatGroq(model = "openai/gpt-oss-20b",temperature = 0)
```

```python
print("For quitting the chat kindly type exit.")
```

```python
user_chat_history_list = []
```

```python
while True:
  question = input("Enter your question:")
  if question.strip().lower() == "exit":
    print("Happy to help you in future")
    break
  docs = retrieve.invoke(question)
  context = "\n\n".join([doc.page_content for doc in docs])
  prompt = f"""you are an AI assistent. user will ask some questions from the given pdf/document. Give the appropriate answers of all the questions
  asked by the user.If user asked any question which have no relation with the pdf then don't provide that question's answer.
  context:
  {context}
  and one more another thing that you have to kept in mind is user's chat history so that if user will ask questions of past history chat then you would
  be able to give the answers of that questions also. So, keep this thing in your memory:{user_chat_history_list}
  question: {question}
  Answer:"""
  response = llm.invoke(prompt)
  print("\nAnswer:")
  print(response.content)
  print("***************************************************************")
  user_chat_history = question + response.content
  user_chat_history_list.append(user_chat_history)
  if len(user_chat_history_list)>3:
    del user_chat_history_list[:-3]
  print(user_chat_history_list)
  print("***************************************************************")
```
